## Defining functions

Functions are used in almost every chunk of code, and you have already been *calling* functions for many different purposes (for example, `print`, `append`, `range`, `type`, `np.arange`). Being able to create a sequence of numbers is so much easier when we can quickly conjure them up with a command like `np.arange(100)`, instead of creating an array from scratch and adding numbers in manually.

Using a function like that is called **calling** a function. This is how we tell the computer to execute function code. Otherwise, the function would exist, but its code would never be executed. In general, the syntax for calling a function goes like `function_name(value1, value2, ...)`, where *each input must be an existing value* that the computer can plug in when executing the code.

In contrast, when we **declare** a new function, the computer does not immediately execute that code. Furthermore, the inputs (or **"arguments"**) in a function definition *do not* need to have pre-existing values. Instead, they can act as placeholder variables which won't be assigned values until the function is **called** with inputs that *do* have values, at which point the function code actually runs for the first time. The table below summarizes this comparison.

| | Declaring a function | Calling a function |
|-|---------------------|--------------------|
|**Syntax**|`def function_name(arg1, arg2):    `<br>`    indented code                   `<br>`    can use arguments here          `<br>`    more indented code              `<br>`    ...                             `|`function_name(value1, value2)`|
|**Value of inputs**| `arg1` and `arg1` have no value, they are <br>only placeholders. | `value1` and `value2` are variables <br>with existing values, or they themselves <br>are values. |
|**Value of function**| `function_name` is an object that can <br>be called. | `function_name(value1, value2)` <br> is evaluated into an output by the computer, <br>which assigns `arg1 = value1` and <br>`arg2 = value2` and then executes the <br>function code. |

In pseudocode, we can assume any given function was **declared** like this:

```
def function(arg1, arg2, ...): 
    implementation
    return output 
```

A great way to see how the functions you use regularly are defined is to look into their documentation. For example, take a look at this documentation for np.arange: https://numpy.org/doc/stable/reference/generated/numpy.arange.html. The very first line looks like this:

```

```

````python
def ekinetic(mass, vel):
    return 0.5 * mass * vel ** 2

KE1 = ekinetic(0.4,10.7)
print("Kinetic energy of this object is", KE1, "joules")

KE2 = ekinetic(0.8,-4.9)
print("Kinetic energy of this object is", KE2, "joules")

KE3 = ekinetic(0.6,-7.4)
print("Kinetic energy of this object is", KE3, "joules")

KE4 = ekinetic(0.5,8.2)
print("Kinetic energy of this object is", KE4, "joules")

print("Total kinetic energy is", KE1 + KE2 + KE3 + KE4, "joules")
````

## Conditional statements

You might be familiar with flowcharts like the one below. They can help us organize decision-making into a logical structure with steps. In the case below, we can decide how to make a lamp work by performing actions sequentially, and using new information to make each sub-decision along the way.

<img src="https://images.edrawmax.com/images/knowledge/flowchart-1-what.png" alt="Flowchart diagram titled 'Lamp doesn't work'. At the top of the flowchart, an arrow leads into 'Lamp plugged in?' No leads to 'Plus in lamp.' Yes leads to 'Bulb burned out?' Yes leads to 'Replace bulb.' No leads to 'Repair lamp.'" width="200"/>

Computers can use a type of flowchart structure when executing code. In Python, we call these structures **conditional statements**, or `if`/`elif`/`else` statements. The diamond-shaped "test expressions" in the diagram below would correspond to the yellow diamonds with yes/no questions in our lamp example.

The main difference between these structures is that instead of asking the computer a yes/no question, we need to pose it a True/False expression (a "Boolean"). For example instead of asking, "is the lamp plugged in?", we would say "True or False, the lamp is plugged in." Then the computer would evaluate whether the statement is True or False, and proceed from there.

To formalize this tested of True/False, here is another way to visualize how the computer processes a conditional statement.

<img src="https://media.geeksforgeeks.org/wp-content/uploads/if-statement.jpg" alt="" width="200"/>

To translate this flowchart into pseudocode format, it would look like this:

```
if test_expression:
    body of if
more code written here
```

When a computer reads this code from top to bottom, it takes the exact steps of following the arrows in the flowchart above. For example, if Test Expression is `True`, then the computer will execute whatever is in "Body of if," and then carry on.
