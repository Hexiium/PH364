````python
import numpy as np

def small_change(A, dim):
    i = np.random.randint(dim)
    j = np.random.randint(dim)
    A[i][j] = np.random.random()
````

````python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

plt.rcParams["animation.html"] = "jshtml" 
plt.rcParams['figure.dpi'] = 100
plt.ioff()
fig, ax = plt.subplots()

# initialize the object being animated
L = 20
grid = np.zeros((L, L))

def animate(frame):
    plt.cla()
    
    # make a small change to the object
    small_change(grid, L)
    
    plt.imshow(grid, cmap="turbo")
    plt.xticks([])
    plt.yticks([])

FuncAnimation(fig, animate, frames=150)
````

## Resolving a push/pull conflict

Sometimes, when you try to `push` or `pull`, all is not well, and you encounter a conflict. This means that in the time since you last had a successful push/pull, you've made some changes locally, and the remote repo has changed too. This can happen if you're preparing a new file locally that you want to push onto the remote repo, but one of your peers already did the same thing and pushed their file first.

This is what that can look like:

<img src="https://raw.githubusercontent.com/hamerski/PH-36X-Public/refs/heads/main/images/divergent-branches.png" 
     alt="Alt text here" 
     width="600"/>

In an earlier class, we told you to use `--rebase` to try to resolve this **(don't do this anymore!)**. Going forward, we'll use a conceptually simpler option, and also safer for not accidentally losing work.

When you see this, run: `git config pull.rebase false`

This sets a default configuration for your `pull` commands. Then, try: `git pull` again.

This time, your terminal could prompt you to enter a merge message. Just like for a commit message, you are simply being asked to explain the change that you are trying to make. When this prompt happens during a pull, one difference is that git creates a default merge message for you, so you don't need to add any text to it. Simply quit the `vi` text editor by hitting the `Esc` key to go into command mode, and then typing `:wq`+`Enter` to save and quit.

From here, you can run a `git push` if you have a commit that you were waiting to push.

So when you try to push a commit and can't, do this instead:

```
git config pull.rebase false
git pull
Esc + :wq + Enter
git push
```

That first `config` line should only need to be run once ever to set your configuration. The `Esc` + `:wq` + `Enter` are the `vi` text editor navigation commands to accept the default merge message.
