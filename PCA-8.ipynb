## Detecting Particles Indirectly

CERN measures hundreds of thousands of collisions, often millions to detect a single particle. Many of the particles have a such a small lifetime, that the detectors themselves will never see them directly. Instead, their existence must be inferred through the products that the short-lived particles decay into.

A useful tool for making these inferences are particle conservation laws. These quantities must all be conserved:
- charge
- spin
- lepton number
- ...many other quantities!

You have worked with charge and spin before. **Lepton number** refers to the leptons (L = 1) and anti-leptons (L = -1) in the interaction or decay. There are three different types of leptons: electron, muons, and tau particles. For example, an electron has $L_e = 1$. In reality, all three lepton numbers must be conserved ($L_e$, $L_\mu$, and $L_\tau$), but for now, we can consider these as just a singular $L$.

## Invariant Mass of Collided Particles

One the most vital conservation laws in particle physics is the conservation of energy-momentum. This considers the invariant mass that can be computed from the energy momentum 4-vector (1 dimension of energy and 3 dimensions of momentum), which includes $E$, $p_x$, $p_y$, and $p_z$.

This comes from adding a momentum term to the energy-mass equivalence equation:

$E = mc^2 + pc$

And then we set c = 1, and treat energy as fourth dimension alongside the usual 3 space dimensions in momentum:

$$m = \sqrt{E^2 - (p_x^2 + p_y^2 + p_z^2)}$$

Like the other conservation laws, this holds across reference frames, meaning that if we can measure the invariant mass from detectors in a lab, that is the same as the invariant mass from the reference frame of the collision.

**&#9989; Do this:** Consider the beta decay once again. An intermediate part of this decay is the emission of a W boson, which is much more massive than a proton or neutron. How is this possible, given the conservation of the invariant mass quantity?

For energy to be conserved, the momentum of the W boson would have to be lower for its mass to exist for its short lifetime.
```python
import pandas as pd

df = pd.read_csv("https://opendata.cern.ch/record/545/files/Jpsimumu.csv")
```

**&#9989; Do this:** Compute the invariant mass and add it as a column to the data frame.

```python
c = 2.9*10**8

df['E'] = df['E1'] + df['E2']
df['px'] = df['px1'] + df['px2']
df['py'] = df['py1'] + df['py2']
df['pz'] = df['pz1'] + df['pz2']

df['inv_m'] = (df['E']**2 - (df['px']**2 + df['py']**2 + df['pz']**2))**0.5

df['inv_m'].describe()
```

**&#9989; Do this:** Produce a histogram visualization that shows the first peak (around 3.1 GeV) very clearly -- this is the same one that was shown in the video.
