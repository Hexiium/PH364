```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/MuRun2010B.csv", skiprows=5)

df['Q_sum'] = df['Q1'] + df['Q2']

df['E'] = df['E1'] + df['E2']
df['px'] = df['px1 '] + df['px2']
df['py'] = df['py1'] + df['py2']
df['pz'] = df['pz1'] + df['pz2']

df['inv_m'] = (df['E']**2 - (df['px']**2 + df['py']**2 + df['pz']**2))**0.5

dat = df[df['Q_sum'] == 0]
noise = df[df['Q_sum'] != 0]

dat['inv_m'].describe()
```

```python
plt.hist(dat['inv_m'],bins=100, label='Invariant Mass (Q=0)',range=(0,10))
plt.hist(noise['inv_m'],bins=100, label='Noise (Q=-2,2)',range=(0,10))
plt.legend()
plt.xlabel('Mass')
plt.ylabel('Count')
plt.title('Muon-Muon Invariant Mass Histogram')
plt.grid()
```

```python
plt.hist(dat['inv_m'],bins=100, label='Invariant Mass',range=(50,110))
plt.hist(noise['inv_m'],bins=100, label='Noise',range=(50,110))
plt.legend()
plt.axis(ymax=75)
plt.xlabel('Mass')
plt.ylabel('Count')
plt.title('Muon-Muon Invariant Mass Histogram')
plt.grid()
```
