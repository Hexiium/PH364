## Part 1: Set up a group GitHub repo

You can find more detailed instructions for this process back on ICA 1.2.

The overview is:
1. One group member create a GitHub repo, and add your members as collaborators.
    - Also add the faculty instructor and TA: `hamerski` and `minifold`
2. Every group member create a directory in your JupyterHub, and initiate a local repo there.
3. Link your local repo (JH) to the remote repo (GH).

Once you are in your newly created directory in your terminal, use these commands to initiate a connection to the GitHub repo and set it to be upstream from your local repo:

```
git init
git remote add origin git@github.com:<user>/<repository>.git
git pull --set-upstream origin main
```

You can see PCA 3.1 for more details about what these commands mean, and how you can pull and push successfully from a shared repo.

You can check your setup with these commands:

- `git remote -v`: You should see two lines with the origin SSH URL, one with fetch and one with push. This means the remote GitHub repo has been set up properly as origin.
- `git branch -vv`: You should a line that starts with, "* main" and also contains "\[origin/main\]." This means you set the remote GitHub repo main branch to be upstream from your main branch on your local repo.

## Part 3: Resources for modeling sound waves

In PCA 4, you learned about a model for the motion of a vibrating string. We are jumping ahead into this modeling cycle after exploring Step 2 in the PCA: making assumptions and gathering resources.

A major resource we are using is a new numerical method called a **finite-difference method.** This relies on a string being discretized in many little equally sized chunks.

<img src="https://raw.githubusercontent.com/hamerski/PH-36X-Public/refs/heads/main/images/FDM-diagram.png" 
     alt="Alt text here" 
     width="500"/>


In the diagram above, each chunk of the string has its own y-value. **Finite-difference methods** allow us to define a discretized version of the second derivative of $y$ with respect to $x$:

$$\frac{\partial^2 y_i}{\partial x^2} = \frac{y_{i+1} - 2y_i + y_{i-1}}{\Delta x^2}$$

An important part of "gathering resources" in Step 2 of the modeling cycle is to determine **how** the resources you have can be applied to the problem at hand.

&#x2705; With your group, discuss how this FDM can be used to model the evolution of the string in motion. 

&#x2705; Apply this to the example of a plucked string. If the string is released at $t = t_0$, use the FDM to compute the position + velocity of all the string's chunks one time-step later, at $t = t_0 + \Delta t$.
