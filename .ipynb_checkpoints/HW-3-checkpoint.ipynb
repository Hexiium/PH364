The models of string vibration and musical instruments you worked on over Weeks 4-6 were based on finite-difference methods and Fourier transforms.

In this homework, you will build on the code you and your group wrote during class to solve some additional problems related to these methods.

You can have some free starter code related to finite-difference methods, since we didn't get a chance to code that method much during class.

````python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.animation import FuncAnimation
````

````python
# free starter functions for finite-difference methods!
# you can use this however you like, or don't :)

# wave equation translated into FDM
def d2y_dt2(y, c, dx):
    deriv = np.zeros_like(y)
    y_minus = y[:-2]
    y_center = y[1:-1]
    y_plus = y[2:]
    
    y_terms = y_minus - 2 * y_center + y_plus
    
    # deriv stays 0 at the endpoints
    deriv[1:-1] = c ** 2 * (y_terms) / dx ** 2
    
    return deriv

# update the entire string by one time step
def string_step(x, y, vel, c, dt):
    # dx for discretized string
    dx = x[1] - x[0]
    
    # d2y / dt2
    accel = d2y_dt2(y, c, dx)

    # need to update y velocity and position
    vel += accel * dt
    y += vel * dt
````

````python
# initialize the plucked string

def pluck_init(x, disp):
    N = len(x)
    dx = x[1] - x[0]
    
    # start with uniform values
    uniform = np.linspace(-1, 1, len(x))
    
    # make flipped absolute value function for pluck shape
    pluck = -1 * np.abs(uniform)
    
    # scale and shift
    y = (pluck + 1) * disp

    return y
````

````python
# initial conditions

L = 1.0
N = 200

x = np.linspace(0, L, N)
dx = x[1] - x[0]

disp = 0.01
freq = 220
````

````python
# animation code

plt.rcParams["animation.html"] = "jshtml"
plt.rcParams['figure.dpi'] = 60
plt.ioff()
fig, ax = plt.subplots()

y = pluck_init(x, disp)
v = np.zeros_like(y)
c = 2 * L * freq

dt = 0.00001

steps_per_frame = 5
y_bound = 1.1 * disp

def animate(t):
    plt.cla()
    
    plt.plot(x, y)
    plt.ylim(-y_bound, y_bound)

    for i in range(steps_per_frame):
        string_step(x, y, v, c, dt)    

FuncAnimation(fig, animate, frames=91)
````
