## Assumptions in Modeling

The first two steps of the modeling cycle are: 
1. Pose a new question about a real-world physics phenomenon
2. Gather your resources and make assumptions to create a simplified "model-world" phenomenon.

Assumptions (step 2) are inseparably linked to the question being asked (step 1). By simplifying the real-world into the model-world, we must choose to ignore some features of the real-world phenomenon. In doing so, we highlight the features that matter, as determined from the question being asked.

For example, consider the Bohr model of an atom, which highlights how electrons can occupy separate energy levels. The concentric circles in this model are a great way to visualize the fact that the separate levels correspond to different energies.

<img src="https://www.physicsclassroom.com/getmedia/7cf3a4a8-df89-426e-b372-bdd5cde9bc1d/Ch3L1b10.png" 
     alt="Bohr model with electron energy levels of an atom" 
     width="300"> 

The choice to draw circles here represents an (unrealistic) assumption that electron orbitals can be represented as concentric circles. This is not realistic when thinking about the actual positions that electrons can occupy with respect to the nucleus. However, this model is good for asking questions about the energies of electrons in their bound states or the band structure of an atom. We would not want to use this model for investigating the probabilistic nature of electrons and their wavefunctions.

Another example of an atomic model is the electron cloud. This is better suited for visualizing the overall distribution of electric charge in an atom. 

<img src="https://www.physicsclassroom.com/getmedia/84cd4371-b79f-4e5e-9cde-3055cbbb95ab/Ch5L2a2.png" 
     alt="Electron cloud model of atom" 
     width="250"> 
     
This model also relies on an unrealistic assumptions -- this time, that electrons occupy a spherically symmetric cloud around a nucleus. This is great for analyzing the atom as a whole with both positive and negative charge present. If we were to ask questions about the behavior of charges in an electrical insulator, this model has everything we need to show how charges could be organized within such a material. However, we would not use this model to investigate questions about energy orbitals.
    

Yet another example of an atomic model is the spherical harmonic orbitals. This showcases the probabilistic position distribution of electrons in a single atomic orbital. 
     
<img src="https://upload.wikimedia.org/wikipedia/commons/c/cf/HAtomOrbitals.png" 
     alt="Spherical harmonic orbitals of a hydrogen atom" 
     width="250"> 
     
Because of the focus on individual orbital shapes that we see in the spherical harmonics, it is difficult to overlay multiple orbitals together. In this model, we choose to model each orbital on it's own, which relies on the unrealistic assumption that electrons exist only in their bound states. This model is well suited for answering questions about individual solutions to electron wavefunctions and individual orbitals. However, it's not great for discussing the behavior of an atomic system containing multiple orbitals, or even a single electron that goes through an excitation from one orbital to another.

## Modeling a magnet

In the recent ICA, we asked, "How hot can a magnet get until it's not a magnet anymore?"

By modeling a magnetic material with mean field theory, we assumed that the behavior of the material could be modeled from the mean behavior of its individual particles. This means that we assumed all the relevant physics for our question could be captured at the scale of individual atoms.

However, mean field theory does not capture the actual physical mechanisms at play. As a model, it would not be well-suited for a question like, "How do atoms in a magnetic material behave at different temperatures?"
