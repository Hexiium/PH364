```python
import plotly.express as px
import pandas as pd
import numpy as np

N=5
grid = np.zeros([N**3,3])

n = 0
for i in range(N):
    for j in range(N):
        for k in range(N):
            grid[n, 0:3] = (i, j, k)
            n += 1
            
df = pd.DataFrame(grid, columns=['x','y','z'])

fig = px.scatter_3d(df, x='x', y='y', z='z')

fig.update_layout(
    scene=dict(
        xaxis=dict(range=[-1, N]),
        yaxis=dict(range=[-1, N]),
        zaxis=dict(range=[-1, N]),
        aspectmode='cube'
    )
)

fig.show()
```
