## 1. Create a GitHub account

In this course, we'll collaborate on several coding projects, from small groups to the whole class. To do so we'll use GitHub, which is the "industry standard" public repository for code. It allows both users and the public to download, execute, and run projects.

The first thing you will need to do is set up an account on [GitHub](https://github.com). 

It is a good idea to set this account up with your private email first, and add your university email as a secondary email later. That way you can manage your github account after you leave. 

Adding your university email allows you to sign up for GitHub Pro, which allows you to make use of more storage and better collaboration tools (not required for today's software setup).

## 2. Open the JupyterHub terminal

In this course, we'll also frequently use a terminal. Through the terminal, we can issue commands directly to the JupyterHub server. Using a terminal is a useful skill to build because the commands we'll learn are the same ones used widely in many different computing contexts. 

To open the terminal, create a new tab within JupyterHub (not a new browser tab). On the right end of your tabs, click `+` and then select `Terminal`.

The terminal provides access to using the command-line interface (CLI). This is where we can type commands. These commands will be interpreted by JupyterHub's "Shell," and then the JupyterHub server will carry out the commands.

## 3. Create an SSH key on JupyterHub

Our next step is to establish a connection between your JupyterHub account and your GitHub account. This will enable you to write and edit code on JupyterHub, and then share those changes with your peers.

A nice way to establish this connection is through SSH ("Secure Shell"), which is a software package that facilitates secure file transfers across different servers. In our case, we want to transfer files between the JupyterHub server and the GitHub server.

In your JupyterHub terminal, type this line and hit enter to issue a command: 

```
ssh-keygen -t ed25519
```

Press `Enter` to accept default values. 
This creates a SSH key and puts it in: `~/.ssh/id_ed25519.pub`.

View the file. You can use this command to do so: 

```
cat ~/.ssh/id_ed25519.pub
```

Copy the **entire line** of text. It should look something like this: `ssh-ed25519 <string of characters> <user@ph364x001>`.

## 4. Add your SSH key to your GitHub account

Finally, you need to provide the information about your SSH key to your GitHub account.

On GitHub, go to `Settings`. You can find them by clicking on your icon in the top right.

Then, on the left sidebar in your settings, navigate to `SSH and GPG Keys` --> `New SSH Key`

Title the new key based on where your SSH key originated, for example "PH 364 JupyterHub."

Paste the **whole line of text** from earlier into the Key textbox, and add the key.
